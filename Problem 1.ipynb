```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

### 1. Create a sample dataset with some missing values

```python
data = {
    "Age": [22, 28, np.nan, 41, 35, 26],
    "Salary": [32000, 48000, 55000, np.nan, 42000, 60000],
    "Department": ["IT", "Finance", "IT", "Marketing", np.nan, "Finance"],
    "Years_of_Experience": [1, 4, 2, 12, 7, np.nan],
}

df = pd.DataFrame(data)
```

### 2. Separate numerical and categorical columns


```python
numeric_features = ["Age", "Salary", "Years_of_Experience"]
categorical_features = ["Department"]
```

### 3. Create preprocessing pipelines

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]
)
```

### 4. Combine both preprocessing pipelines

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
    ]
)
```

### 5. Fit and transform the dataset

```python
processed_data = preprocessor.fit_transform(df)

print("--- Original Dataset ---")
print(df)

print("\n--- Processed Feature Matrix ---")
print(processed_data)
```
