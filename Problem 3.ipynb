```python
import pandas as pd
```

### Create the DataFrame

```python
data = {
    "Student Name": ["Arjun", "Mita", "Rohan", "Ananya", "Karan"],
    "Roll Number": [201, 202, 203, 204, 205],
    "Marks": [88, 76, 93, 64, 81],
    "Attendance": [87, 90, 94, 78, 89]
}

df = pd.DataFrame(data)
```

### Function to calculate grade

```python
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"
```

### Add the dynamically calculated Grade column

```python
df["Grade"] = df["Marks"].apply(calculate_grade)
```

### Display the DataFrame

```python
print(df)
```
