```python
import pandas as pd
```

### Create the DataFrame

```python
data = {
    "Student Name": ["Arjun", "Mita", "Rohan", "Ananya", "Karan"],
    "Roll Number": [201, 202, 203, 204, 205],
    "Marks": [74, 89, 81, 65, 94],
    "Attendance": [88, 91, 84, 79, 96]
}

df = pd.DataFrame(data)
```

### Filter students who scored above 80 marks

```python
filtered_df = df[df["Marks"] > 80]
```

### Display the filtered records

```python
print("Students who scored above 80 marks:")
print(filtered_df)
```
