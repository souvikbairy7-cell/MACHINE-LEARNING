# problem 1

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
```

### 1. Load the Wine Dataset

```python
wine_data = load_wine()
```

### Create a DataFrame

```python
wine_df = pd.DataFrame(
    wine_data.data,
    columns=wine_data.feature_names
)
```

### Add target column

```python
wine_df["target"] = wine_data.target
```

### 2. Basic Dataset Exploration

```python

print("\n========== FIRST FIVE ROWS ==========")
print(wine_df.head())

print("\n========== LAST FIVE ROWS ==========")
print(wine_df.tail())

print("\n========== DATASET SHAPE ==========")
print("Rows and Columns:", wine_df.shape)

print("\n========== DATASET INFORMATION ==========")
wine_df.info()

print("\n========== COLUMN NAMES ==========")
print(wine_df.columns.tolist())
```

### 3. Statistical Analysis

```python
print("\n========== STATISTICAL SUMMARY ==========")
print(wine_df.describe())
```

### 4. Missing Values and Duplicates

```python

print("\n========== MISSING VALUES ==========")
print(wine_df.isnull().sum())

print("\n========== TOTAL MISSING VALUES ==========")
print(wine_df.isnull().sum().sum())

print("\n========== DUPLICATE ROWS ==========")
print(wine_df.duplicated().sum())
```

### 5. Target Class Analysis

```python
print("\n========== TARGET CLASS COUNT ==========")
print(wine_df["target"].value_counts().sort_index())
```

### 6. Data Types and Unique Values

```python
print("\n========== DATA TYPES ==========")
print(wine_df.dtypes)

print("\n========== UNIQUE VALUES ==========")
print(wine_df.nunique())
```

### 7. Correlation Analysis

```python
correlation = wine_df.corr(numeric_only=True)

print("\n========== CORRELATION MATRIX ==========")
print(correlation)
```

### 8. Correlation Heatmap


```python
plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Wine Dataset")
plt.tight_layout()
plt.show()
```

### 9. Histograms - Feature Distribution

```python
wine_df.hist(
    figsize=(16, 12),
    bins=15
)

plt.suptitle(
    "Distribution of Wine Dataset Features",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

### 10. Boxplot - Outlier Detection

```python
plt.figure(figsize=(16, 8))

sns.boxplot(
    data=wine_df.drop(columns="target")
)

plt.title("Boxplot for Detecting Outliers")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### 11. Scatter Plot

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=wine_df,
    x="alcohol",
    y="malic_acid",
    hue="target"
)

plt.title("Alcohol vs Malic Acid")
plt.xlabel("Alcohol")
plt.ylabel("Malic Acid")
plt.tight_layout()
plt.show()
```

### 12. Pairplot

```python
sns.pairplot(
    wine_df,
    vars=[
        "alcohol",
        "malic_acid",
        "color_intensity",
        "proline"
    ],
    hue="target"
)

plt.suptitle(
    "Pairwise Relationship of Selected Wine Features",
    y=1.02
)

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=wine_df,
    x="target"
)

plt.title("Distribution of Wine Target Classes")
plt.xlabel("Wine Class")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()
```
