```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
```

### Load Wine dataset


```python
wine_data = load_wine()
```

### Create DataFrame


```python
wine_df = pd.DataFrame(
    wine_data.data,
    columns=wine_data.feature_names
)
```

### # Calculate correlation matrix

```python

corr = wine_df.corr()
```

### Display correlation matrix

```python
print("\n--- Correlation Matrix ---")
print(corr)
```

###  Generate correlation heatmap

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Wine Dataset")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Copy correlation matrix


```python
corr_copy = corr.copy()
```

### # Remove diagonal values

```python
for i in range(len(corr_copy)):
    corr_copy.iloc[i, i] = -1

# Find strongest positive correlation
feature_pair = corr_copy.stack().idxmax()
highest_corr = corr_copy.stack().max()

print("\n--- Strongest Positive Correlation ---")
print("Feature 1:", feature_pair[0])
print("Feature 2:", feature_pair[1])
print("Correlation:", highest_corr)
```
